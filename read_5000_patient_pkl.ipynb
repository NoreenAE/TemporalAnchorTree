# Read 5000-Patient PKL Files

This notebook reads the active 5000-patient synthetic data and labels PKL files and prints basic information.

```python
from pathlib import Path

import pandas as pd

data_path = Path("synthetic_temporal_unit_test_5000.pkl")
labels_path = Path("synthetic_temporal_unit_test_labels_5000.pkl")

data = pd.read_pickle(data_path)
labels = pd.read_pickle(labels_path)

print("Data file:", data_path)
print("Data shape:", data.shape)
print("Data columns:", data.columns.tolist())
print("Unique patients in data:", data["patient_id"].nunique())

print("\nLabels file:", labels_path)
print("Labels shape:", labels.shape)
print("Labels columns:", labels.columns.tolist())
print("Unique patients in labels:", labels["patient_id"].nunique())

display(data.head())
display(labels.head())
```
